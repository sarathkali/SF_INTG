# Account-Wide Tagging: Business Domain + Sensitive Data Classification

```
%%sql -r use_role
USE ROLE ACCOUNTADMIN;
```

```
%%sql -r create_schema
CREATE DATABASE IF NOT EXISTS GOVERNANCE_DB;
CREATE SCHEMA IF NOT EXISTS GOVERNANCE_DB.TAGS;
```

```
%%sql -r create_tags
CREATE OR REPLACE TAG GOVERNANCE_DB.TAGS.CONFIDENTIALITY
  ALLOWED_VALUES 'PUBLIC', 'INTERNAL', 'CONFIDENTIAL', 'RESTRICTED'
  COMMENT = 'Data confidentiality level';

CREATE OR REPLACE TAG GOVERNANCE_DB.TAGS.DOMAIN
  ALLOWED_VALUES 'FINANCE', 'RETAIL', 'CALL_CENTER', 'ENGINEERING', 'GENERAL'
  COMMENT = 'Business domain classification';

CREATE OR REPLACE TAG GOVERNANCE_DB.TAGS.DATA_LAYER
  ALLOWED_VALUES 'RAW', 'STAGING', 'ANALYTICS', 'PRESENTATION'
  COMMENT = 'Data layer in the pipeline';

CREATE OR REPLACE TAG GOVERNANCE_DB.TAGS.OWNER
  COMMENT = 'Team or person owning the data asset';
```

```
%%sql -r tag_call_center
ALTER TABLE CALL_CENTER_DB.IMAGES.CORTEX_AI_USAGE_AUDIT SET TAG
  GOVERNANCE_DB.TAGS.DOMAIN = 'CALL_CENTER',
  GOVERNANCE_DB.TAGS.DATA_LAYER = 'ANALYTICS',
  GOVERNANCE_DB.TAGS.CONFIDENTIALITY = 'INTERNAL';

ALTER TABLE CALL_CENTER_DB.IMAGES.PDF_CHUNKS SET TAG
  GOVERNANCE_DB.TAGS.DOMAIN = 'CALL_CENTER',
  GOVERNANCE_DB.TAGS.DATA_LAYER = 'STAGING',
  GOVERNANCE_DB.TAGS.CONFIDENTIALITY = 'INTERNAL';

ALTER TABLE CALL_CENTER_DB.RAW.AUDIO_FILES SET TAG
  GOVERNANCE_DB.TAGS.DOMAIN = 'CALL_CENTER',
  GOVERNANCE_DB.TAGS.DATA_LAYER = 'RAW',
  GOVERNANCE_DB.TAGS.CONFIDENTIALITY = 'CONFIDENTIAL';

ALTER TABLE CALL_CENTER_DB.STAGE.TRANSCRIBE_AUDIO_FILES SET TAG
  GOVERNANCE_DB.TAGS.DOMAIN = 'CALL_CENTER',
  GOVERNANCE_DB.TAGS.DATA_LAYER = 'STAGING',
  GOVERNANCE_DB.TAGS.CONFIDENTIALITY = 'CONFIDENTIAL';
```

```
%%sql -r verify_business_tags
SELECT SYSTEM$GET_TAG('GOVERNANCE_DB.TAGS.DOMAIN', 'TEST_DB.PUBLIC.EMPLOYEE', 'TABLE') AS DOMAIN_TAG,
       SYSTEM$GET_TAG('GOVERNANCE_DB.TAGS.DATA_LAYER', 'TEST_DB.PUBLIC.EMPLOYEE', 'TABLE') AS DATA_LAYER_TAG,
       SYSTEM$GET_TAG('GOVERNANCE_DB.TAGS.CONFIDENTIALITY', 'TEST_DB.PUBLIC.EMPLOYEE', 'TABLE') AS CONFIDENTIALITY_TAG;
```

```
%%sql -r verify_pii_tags
SELECT SYSTEM$GET_TAG('SNOWFLAKE.CORE.SEMANTIC_CATEGORY', 'TEST_DB.PUBLIC.EMPLOYEE.ENAME', 'COLUMN') AS SEMANTIC_TAG,
       SYSTEM$GET_TAG('SNOWFLAKE.CORE.PRIVACY_CATEGORY', 'TEST_DB.PUBLIC.EMPLOYEE.ENAME', 'COLUMN') AS PRIVACY_TAG;
```

```
%%sql -r full_tag_report
SELECT OBJECT_DATABASE, OBJECT_SCHEMA, OBJECT_NAME, COLUMN_NAME, TAG_NAME, TAG_VALUE
FROM SNOWFLAKE.ACCOUNT_USAGE.TAG_REFERENCES
WHERE OBJECT_DATABASE NOT IN ('SNOWFLAKE', 'SNOWFLAKE_SAMPLE_DATA')
ORDER BY OBJECT_DATABASE, OBJECT_SCHEMA, OBJECT_NAME, TAG_NAME;
```
